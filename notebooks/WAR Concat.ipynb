```python
import pandas as pd

# Load your Excel file
excel_file = "TeamWAR.xlsx"  # <- Change to your actual file path
xl = pd.ExcelFile(excel_file)

batting_sheets = []
pitching_sheets = []

for sheet in xl.sheet_names:
    # Extract the year from the sheet name (e.g., 'Bat_2025' -> 2025)
    year = int(sheet.split("_")[1])

    # Read the sheet into a dataframe
    df = pd.read_excel(excel_file, sheet_name=sheet)

    # Add a Year column so we know which season this data belongs to
    df["Year"] = year

    if sheet.startswith("Bat_"):
        batting_sheets.append(df)
    elif sheet.startswith("Pitch_"):
        pitching_sheets.append(df)

# Combine all years into master DataFrames
all_batting = pd.concat(batting_sheets, ignore_index=True)
all_pitching = pd.concat(pitching_sheets, ignore_index=True)

# Export to 2 clean CSV files
all_batting.to_csv("all_batting.csv", index=False)
all_pitching.to_csv("all_pitching.csv", index=False)

print("Master CSVs created successfully!")
```
